çoklu doğrusal regresyon/ eksik veri değerlerini doldurarak çoklu doğrusal regresyon analizi yapımı.

```python
import pandas as pd
from sklearn.impute import SimpleImputer
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import sklearn.metrics as mt
```

```python
df=pd.read_csv("D:/projects/python_tum_dosyalar/data-science-lab/ae_makina_ogrenmesi/16_basit_dogrusal_regresyon/data/advertising.csv")
columns = ["TV", "Radio"]

for col in columns:
    random_index = df[col].sample(n=6).index
    df.loc[random_index, col] = None
df.to_csv("../data/eksik_deger_veri.csv",index=False)
```

csv dosyasında eksik değer oluşturduk


```python
veri=pd.DataFrame(pd.read_csv("../data/eksik_deger_veri.csv"))
veri.isnull().sum()
```

```python
columns=["TV","Radio"]
imputer=SimpleImputer(missing_values=np.nan,strategy="mean")
imputer=imputer.fit(veri[columns])
veri[columns]=imputer.transform(veri[columns])
veri=veri[["TV","Radio","Sales"]]
veri.isnull().sum()
```

eksik verileri ortalama ile doldurduk

```python
y=veri["Sales"]
x=veri[["TV", "Radio"]]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
lr=LinearRegression()
lr.fit(x_train,y_train)
tahmin=lr.predict(x_test)
```

çoklu doğrusal regresyon yaptık

```python
r2=mt.r2_score(y_test,tahmin)
mse=mt.mean_squared_error(y_test,tahmin)
rmse=mt.root_mean_squared_error(y_test,tahmin)
mae=mt.mean_absolute_error(y_test,tahmin)
print(r2,mse,rmse,mae)
```
