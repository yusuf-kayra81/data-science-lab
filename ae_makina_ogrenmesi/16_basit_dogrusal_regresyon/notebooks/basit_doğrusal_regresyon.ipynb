```python
import numpy as np
import pandas as pd

# Data Visualisation
import matplotlib.pyplot as plt 
import seaborn as sns
import statsmodels.api as sm
from sklearn.linear_model import LinearRegression
```

```python
advertising = pd.DataFrame(pd.read_csv("D:/projects/python_tum_dosyalar/data-science-lab/ae_makina_ogrenmesi/16_basit_dogrusal_regresyon/data/advertising.csv"))
advertising.head()
```

```python
tv=advertising[["TV"]]
sales=advertising[["Sales"]]
plt.scatter(tv,sales)
plt.show()


```

```python
sabit=sm.add_constant(tv)
model=sm.OLS(sales,sabit).fit()
model.summary()
```

```python
lr=LinearRegression()
lr.fit(tv,sales)
lr.coef_,lr.intercept_
```
